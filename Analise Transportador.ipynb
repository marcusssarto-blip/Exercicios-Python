```python
import pandas as pd
import numpy as np
```

```python
#Criando Dicionário

transportadoras = {
    "JSL": 10,
    "TAG": 15,
    "Figma": 25,
}

print(transportadoras)
```

```python
# Criando Dataframe

dados_agosto = {
    "Mês": ["Agosto","Agosto","Agosto","Agosto","Agosto","Agosto","Agosto", "Agosto"],
    "Cliente": ["Pernambucanas",None,"Lojas 100","Casas Bahia","Amazon","MELI","LoJas 100", "C_Bahia"],
    "Transportador": ["JSL","TAG","Figma",None,"tag","Jsl","FIGMA","TaG"],
    "Peso": [25,10,20,None,85,30,5,1000],
    "Distância (Km)": [60,15,32,1000,95,20,34,None],
    "Dias_Entrega": [2,5,7,8,15,30,56,72],
    "Status": ["Atrasado","atraAsaso",None,"intregue","Atrasdo","entregue","Entregue","AtraADO"]
}

df_agosto = pd.DataFrame(dados_agosto)
display(df_agosto)

dados_setembro = {
    "Mês": ["Setembro","Setembro","Setembro","Setembro","Setembro","Setembro","Setembro", "Setembro"],
    "Cliente": ["Pernambucanas","Colombo","Lojas 100","Casas Bahia","Amazon","MELI","LoJas 100", "C_Bahia"],
    "Transportador": ["JSL","TAG","Figma",None,"tag","Jsl","FIGMA","TaG"],
    "Peso": [60,10,75,24,45,65,30,50],
    "Distância (Km)": [30,14,20,65,75,64,2,64],
    "Dias_Entrega": [14,20,4,9,8,10,6,4],
    "Status": ["Entregue","atraasado"," Entregue","Atrsdo","entrge","entregue","Entregue","AtraADO"]
}

df_setembro = pd.DataFrame(dados_setembro)
display(df_setembro)

df_geral = pd.concat([df_agosto,df_setembro], axis = 0)
df_geral = df_geral.reset_index(drop=True)
display(df_geral)
```

```python
display(df_geral.info())
display(df_geral.describe())
display(df_geral)
```

```python
#Tratamento de Info

df_geral["Cliente"]=df_geral["Cliente"].str.strip()
df_geral["Cliente"]=df_geral["Cliente"].str.title()
df_geral["Status"]=df_geral["Status"].str.strip()
df_geral["Status"]=df_geral["Status"].str.title()
df_geral["Transportador"]=df_geral["Transportador"].replace({
    "Tag": "TAG",
    "tag": "TAG",
    "TaG": "TAG",
    "jsl": "JSL",
    "Jsl": "JSL"
})
df_geral["Cliente"]=df_geral["Cliente"].replace("C_Bahia", "Casas Bahia")
df_geral["Status"]=df_geral["Status"].replace({
    "Atraasaso": "Atrasado",
    "Atrasdo": "Atrasado",
    "Atraado": "Atrasado",
    "Atraasado": "Atrasado",
    "Atrsdo": "Atrasado",
    "Intregue": "Entregue",
    "Entrge": "Entregue",
    })
df_geral["Transportador"]=df_geral["Transportador"].replace("FIGMA", "Figma")
df_geral["Status"]=df_geral["Status"].fillna("Entregue")
df_geral["Cliente"]=df_geral["Cliente"].fillna("Meli")
df_geral["Transportador"]=df_geral["Transportador"].fillna("Figma")
df_geral["Peso"]=df_geral["Peso"].fillna(30)
df_geral["Distância (Km)"]=df_geral["Distância (Km)"].fillna(50)
df_geral = df_geral.rename(columns={"Dias_Entrega":"Lead Time"})
display(df_geral)

display(df_geral.info())
```

```python
#Nova Coluna

df_geral["Frete"] = df_geral["Peso"] * df_geral["Distância (Km)"]

display(df_geral)
```

```python
# Map para SLA Contratado

df_geral["SLA Contratado"] = df_geral["Transportador"].map(transportadoras)


display(df_geral)
```

```python
#Coluna Alerta

df_geral["revisar_alerta"] = (df_geral["Alerta"]=="Revisar").astype(int)
df_geral["Alerta"] = np.where(
    df_geral["Status"] == "Atrasado", "Revisar",
    np.where(df_geral["Lead Time"] > df_geral["SLA Contratado"], "Revisar", "Ok")
)

display(df_geral)
```

```python
#Agrupar e Criar Boolenao do Alerta

df_geral["revisar_alerta"] = (df_geral["Alerta"]=="Revisar").astype(int)


resumo_transportadora = df_geral.groupby("Transportador").agg(
    frete_medio=("Frete", "mean"),
    qtd_entregas=("Cliente", "count"),
    alerta=("revisar_alerta", "sum")
).reset_index()

display(resumo_transportadora)
```

```python
#Renomear colunas e exportar

resumo_transportadora=resumo_transportadora.rename(columns={"frete_medio": "Frete Medio"})
resumo_transportadora=resumo_transportadora.rename(columns={"qtd_entregas": "Qtd Entrega"})
resumo_transportadora=resumo_transportadora.rename(columns={"alerta": "Alerta"})

resumo_transportadora.to_csv("Resumo Transportador.csv",index=False)

display(resumo_transportadora)
```
